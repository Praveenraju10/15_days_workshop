```python
import pandas as pd
```

```python
df=pd.read_csv('student_performance.csv')
```

```python
df.head()
```

```python
print("total students:",df.shape[0])
print("total columns:",df.shape[1])
print("total department:",len(df['department'].unique()))
```

```python
print(df['department'].value_counts())
```

```python
print("Highes maths",df['math_score'].max())
print("Lowest maths",df['math_score'].min())
print("Average maths",df['math_score'].mean().round(2))
print("Average programming score",df['programming_score'].mean())
print(f"Average Attendance,{df['attendance_percentage'].mean()}%")
```

```python
df['total_score']=df[['english_score','math_score','programming_score','science_score']].sum(axis=1)
gender_score=df.groupby('gender')['total_score'].mean().round(2)
print(gender_score)
```

```python
low_att=df[df['attendance_percentage']<75][['name','department','attendance_percentage']]
if len(low_att)>0:
  print(low_att.to_string(index=False))

else:
  print("All students have attendanace above 75%")
```
