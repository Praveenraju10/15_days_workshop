```python
dfo=df.copy()
print(f'copy created:{dfo.shape}')
```

```python
#1. handle the missing values
print('Before fixing nulls:',df.isnull().sum().sum(),'total missing values')

dfo['customer_name'].fillna('unknown customer',inplace=True)
median_qty=dfo['quantity'].median()
dfo['quantity'].fillna(median_qty,inplace=True)
print(f'Filled missing quality with median:{median_qty}')

dfo['category'].fillna('uncategorized',inplace=True)

print('After fixing nulls:',dfo.isnull().sum().sum(),'total missing values')
```

```python
dfo
```

```python
print(f'Before deduplication:{len(dfo)}')
print(f'Duplicate rows:{dfo.duplicated().sum()}')
print('\n Duplicate rows')
print(dfo[dfo.duplicated(keep=False)][['order_id','customer_name','product','order_date']].head(6))
dfo.drop_duplicates(inplace=True)
print(f'\nAfter deduplication:{len(dfo)} rows')
print(f'row reomved:{len(df)- len(dfo)}')
```

```python

dfo['product'].fillna('unknown product',inplace=True)
print(dfo['product'])
print(dfo['product'].isnull().sum())
```

```python
dfo
```

```python
dfo['order_date']=pd.to_datetime(dfo['order_date'],dayfirst=False,errors='coerce')
dfo['order_date'].unique()
```

```python
dfo['year']=dfo['order_date'].dt.year
dfo['month']=dfo['order_date'].dt.month
dfo['day']=dfo['order_date'].dt.day
```

```python
dfo['customer_name'].unique()[:10]
```

```python
dfo['customer_name']=(dfo['customer_name'].str.strip().str.title())
dfo['customer_name'].unique()[:10]
```

```python
mask=(dfo['product']=='keybord') &(dfo['category']=='Electronics')
dfo.loc[mask,'category']='Accessaries'
print(dfo.loc[mask, 'category'])
```

```python
dfo['quantity']=pd.to_numeric(dfo['quantity'],errors='coerce')
dfo['unit_price']=pd.to_numeric(dfo['unit_price'],errors='coerce')
dfo['Revenue']=df['unit_price']*dfo['quantity']
print(dfo[['customer_name','product','quantity','unit_price','Revenue']].head(5))
```

```python
total=dfo['Revenue'].sum()
print(total)
```

```python
print(f'original rows: {len(df)}')
print(f'cleaned rows : {len(dfo)}')
print(f'rows removed : {len(df)-len(dfo)}')
print(f'Missing values: {dfo.isnull().sum().sum()}')
print(f'Duplicate rows: {dfo.duplicated().sum()}')
print(f'Date nulls : {dfo['order_date'].isnull().sum()}')
print(f'Revenue null: {dfo['Revenue'].isnull().sum()}')
```
