```python
df['total_score']=df[['english_score','math_score','programming_score','science_score']].sum(axis=1)
top_performers_index=df['top_performer'].sort_values(ascending=False).index

print(df.loc[top_performers_index,['name','department']])
```

```python
dept_avg=df.groupby('department')['total_score'].mean().round(2)
print(dept_avg.sort_values(ascending=False))
```
